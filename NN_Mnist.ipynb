```python
import torch
import torchvision
from torch.utils.data import DataLoader
import torchvision.transforms as tranforms
import torch.nn as nn
```

```python
device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
```

```python
train=torchvision.datasets.MNIST(root="./data",train=True,transform=tranforms.ToTensor())
test=torchvision.datasets.MNIST(root="./data",train=False,transform=tranforms.ToTensor())
```

```python
train_loader=DataLoader(train,batch_size=64,shuffle=True)
```

```python
#architueture
class NN(nn.Module):
    def __init__(self):
        super(NN,self).__init__()
        self.net=nn.Sequential(
            nn.Linear(784,196),
            nn.ReLU(),
            nn.Linear(196,24),
            nn.ReLU(),
            nn.Linear(24,10)

        )

    def forward(self,x):
        return self.net(x)



```

```python
Model=NN().to(device)
criterion=nn.CrossEntropyLoss()
optimizer=torch.optim.Adam(Model.parameters(),lr=0.001)
```

```python
#training looop

for epoch in range(5):
    for X,y in train_loader:
        X = X.view(X.shape[0], -1).to(device)
        y = y.to(device)

        optimizer.zero_grad()
        outputs=Model(X)

        loss=criterion(outputs,y)
        loss.backward()

        optimizer.step()


    print(f"Epoch {epoch+1}, Loss: {loss.item()}")
print("all done")
```

```python
#test loop
test_load=DataLoader(test,batch_size=64,shuffle=True)


correct=0
total=0
for X,y in test_load:
     X = X.view(X.shape[0], -1).to(device)
     y = y.to(device)

     output=Model(X)
     predictions = output.argmax(dim=1)
     correct += (predictions == y).sum().item()
     total += y.size(0)


print(f"Test Accuracy: {100 * correct / total:.2f}%")
print(total)
```
